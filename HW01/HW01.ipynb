````python

def collect_data(course_urls_file: str) -> pd.DataFrame:
    """
    Collect data from a list of URLs provided in a file.

    Parameters
    ----------
    course_urls_file : str
        Path to the file containing URLs (e.g., 'Lab01/pokemon.txt').

    Returns
    -------
    pd.DataFrame
        A DataFrame containing the required data fields.
    """
    # Load URLs from file
    with open(course_urls_file, 'r') as file:
        urls = [line.strip() for line in file]

    # Initialize empty lists to store the values of each attribute
    sku, names, prices, in_stocks, categories, tags = [], [], [], [], [], []

    # YOUR CODE HERE

    def extract_product_information(url):
        response = requests.get(url)
        soup = BeautifulSoup(response.text, 'html.parser')
        
        summary_section = soup.find('div', class_='summary entry-summary')
        
        sku_element = summary_section.find('span', class_='sku')
        product_sku = int(sku_element.text) if sku_element else None
        
        name_element = summary_section.find('h1', class_='product_title entry-title')
        product_name = name_element.text if name_element else "Unknown Product"
        
        price_element = summary_section.find('span', class_='woocommerce-Price-amount amount')
        product_price = float(price_element.text[1:]) if price_element else 0.0
        
        stock_element = summary_section.find('p', class_='stock in-stock')
        stock_text = stock_element.text if stock_element else "0 in stock"
        product_stock = int(stock_text.replace(' in stock', '').strip()) if 'in stock' in stock_text else 0
        
        categories_element = summary_section.find('span', class_='posted_in')
        product_categories = ', '.join([cat.text for cat in categories_element.find_all('a')]) if categories_element else "No Categories"
        
        tags_element = summary_section.find('span', class_='tagged_as')
        product_tags = ', '.join([tag.text for tag in tags_element.find_all('a')]) if tags_element else "No Tags"
        
        return (product_sku, product_name, product_price, product_stock, product_categories, product_tags)

    craw_data = []
    for url in tqdm(urls):
        craw_data += [extract_product_information(url)]
        
    sku, names, prices, in_stocks, categories, tags = zip(*craw_data)
    # Create DataFrame with collected data
    data = pd.DataFrame({
        "SKU": sku,
        "Name": names,
        "Price": prices,
        "InStock": in_stocks,
        "Categories": categories,
        "Tags": tags
    })

    return data
````

````python
BASE_URL = "https://api.worldbank.org/v2/"
COUNTRIES = ["US", "IN", "CN", "JP", "CA", "GB", "ZA"]
INDICATORS = [
    "SP.POP.TOTL",
    "SP.POP.TOTL.FE.IN",
    "SP.POP.TOTL.MA.IN",
    "SP.DYN.CBRT.IN",
    "SP.DYN.CDRT.IN",
    "SP.DYN.LE00.MA.IN",
    "SP.DYN.LE00.FE.IN",
    "SE.PRM.ENRR",
    "SE.TER.ENRR",
    "SE.PRM.CMPT.ZS",
    "SE.ADT.1524.LT.ZS",
]

# YOUR CODE HERE (option)
# If you need other initializations
countryMap = {
    "us": "United States",
    "in": "India",
    "cn": "China",
    "jp": "Japan",
    "ca": "Canada",
    "gb": "Great Britain",
    "za": "South Africa",
}

featureMap = {
    "SP.POP.TOTL": "Total Population",
    "SP.POP.TOTL.FE.IN": "Female Population",
    "SP.POP.TOTL.MA.IN": "Male Population",
    "SP.DYN.CBRT.IN": "Birth Rate",
    "SP.DYN.CDRT.IN": "Death Rate",
    "SP.DYN.LE00.MA.IN": "Life Expectancy Male",
    "SP.DYN.LE00.FE.IN": "Life Expectancy Female",
    "SE.PRM.ENRR": "Primary School Enrollment",
    "SE.TER.ENRR": "High School Enrollment",
    "SE.PRM.CMPT.ZS": "Primary Completion Rate",
    "SE.ADT.1524.LT.ZS": "Literacy rate"
}
````

````python
def collect_data(country_code: str, per_page: int, start_year: int, end_year: int, max_retries: int = 3, delay: int = 2) -> pd.DataFrame:
    """
    Collect data from the World Bank API for a specific country and date range.

    Parameters
    ----------
    country_code : str
        The ISO 3166-1 alpha-2 code for the country (e.g., 'US' for the United States).
    per_page : int
        Number of records to return per page.
    start_year : int
        The starting year of the data range.
    end_year : int
        The ending year of the data range.
    max_retries : int, optional
        Maximum number of retries for API requests in case of server errors (default is 3).
    delay : int, optional
        Delay (in seconds) between retries (default is 2).

    Returns
    -------
    pd.DataFrame
        A DataFrame containing the data collected from the API.

    Raises
    ------
    ValueError
        If the API request fails or if the data is not available.
    """
    # YOUR CODE HERE
    
    data = []
    
    # Get the country name using the country code
    country_name = countryMap.get(country_code.lower(), "Unknown Country")

    # Loop through each year in the specified range
    for year in range(start_year, end_year + 1):
        # Create a row for the current year and country
        row = {"Year": year, "Country": country_name}

        # Loop through each indicator to fetch its data
        for indicator in INDICATORS:
            url = f"http://api.worldbank.org/v2/country/{country_code}/indicator/{indicator}?format=json&date={year}"
            attempt = 0
            
            while attempt < max_retries:
                try:
                    # Make the API request
                    response = requests.get(url)
                    response.raise_for_status()  # Check for request errors
                    
                    json_data = response.json()
                    if len(json_data) > 1 and json_data[1]:
                        # Extract the value from the response
                        value = json_data[1][0].get('value', None)
                    else:
                        value = None
                    
                    # Store the value in the row
                    row[featureMap[indicator]] = value
                    break  # Exit retry loop if successful
                except requests.exceptions.RequestException:
                    attempt += 1
                    if attempt < max_retries:
                        print(f"Error fetching {indicator} for {year}. Retrying in {delay} seconds...")
                        time.sleep(delay)  # Wait before retrying
                    else:
                        print(f"Failed to fetch {indicator} for {year} after {max_retries} attempts.")
                        row[featureMap[indicator]] = None  # Assign None if failed after retries

        # Add the completed row to the data list
        data.append(row)
    
    # Check if any data was collected
    if not data:
        raise ValueError(f"No data available for country {country_code}")
    
    # Convert the data list into a DataFrame
    df = pd.DataFrame(data)
    return df

    
    
````

````python
def generate_countries_dataset(country_code_list: list) -> pd.DataFrame:
    """
    Generate a dataset by collecting data for a list of countries.

    Parameters
    ----------
    country_code_list : list of str
        List of ISO 3166-1 alpha-2 country codes (e.g., ['US', 'IN', 'CN']).

    Returns
    -------
    pd.DataFrame
        A DataFrame containing the combined data for all countries.

    Raises
    ------
    ValueError
        If data collection for any country fails.
    """
    # Initialize an empty list to store DataFrames
    data_frames = []

    for country_code in country_code_list:
        try:
            # Collect data for each country and append the result to the list
            df = collect_data(country_code=country_code,
                              per_page=100, start_year=2000, end_year=2022)
            data_frames.append(df)
        except ValueError as e:
            print(f"Error collecting data for {country_code}: {e}")

    # Concatenate all collected DataFrames into a single DataFrame
    combined_data = pd.concat(data_frames, ignore_index=True)

    return combined_data
````

````python
def scrape_journal_first_n_page(n):
    total_pages = n
    all_articles = []

    for page in range(1, total_pages + 1):
        BASE_URL = f'https://link.springer.com/journal/42979/articles?filterOpenAccess=false&page={page}'
        response = requests.get(BASE_URL)
        soup = BeautifulSoup(response.text, 'html.parser')

        for article in soup.find_all('article', class_='app-card-open app-card-open--has-image'):
            title = article.find('h3', class_="app-card-open__heading").get_text(strip=True)
            date = get_date(article)
            link = get_link(article)

            new_response = requests.get(link)
            new_soup = BeautifulSoup(new_response.text, 'html.parser')
            authors = get_author(new_soup)
            affiliations = get_affiliations(new_soup)

            all_articles.append([title, date, link, authors, affiliations])

    df = pd.DataFrame(all_articles, columns=['Title', 'Date', 'Link', 'Authors', 'Affiliations'])
    df.to_csv("student_journal.csv", sep='\t', encoding='utf-8', index=False, header=True)

def get_date(article):
    months = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
    meta_div = article.find('div', class_='app-card-open__meta')
    if meta_div:
        tags = meta_div.find_all('span', class_='c-meta__item')
        for tag in tags:
            text = tag.get_text(strip=True)
            if "Open access" in text:
                continue
            if any(month in text for month in months):
                return text
    return None

def get_link(article):
    article_link = article.find('h3', class_="app-card-open__heading").find('a')
    return article_link['href'] if article_link else None

def get_author(new_soup):
    authors = new_soup.find_all('li', class_='c-article-author-list__item')
    author_names = [author.find('a').get_text(strip=True) for author in authors if author.find('a').get_text(strip=True) not in ["…", ""]]
    return ', '.join(author_names[:-1]) + " & " + author_names[-1] if len(author_names) > 1 else author_names[0] if author_names else None

def get_affiliations(new_soup):
    affiliations = new_soup.find_all('li', id=lambda x: x and x.startswith('Aff'))
    return "; ".join(f"{aff.find('p', class_='c-article-author-affiliation__address').get_text(strip=True)}: {aff.find('p', class_='c-article-author-affiliation__authors-list').get_text(strip=True)}" for aff in affiliations) if affiliations else None


scrape_journal_first_n_page(1)
student_journal = pd.read_csv("./student_journal.csv", sep="\t")
student_journal
````
